```python
df=pd.read_csv("titanic.csv")
```

```python
x=df["Age"]                                     #x-axis
```

```python
marks=[40,50,70,90,80]
classss=["A","B","C","D","E"]
```

```python
plt.bar(classss,marks)                  #Bar Graph - X-Catagorical vs Y-Numerical
plt.xlabel("Class")
plt.ylabel("Marks")
plt.title("Class Vs Marks")
```
