```python
a=np.arange(5)                          #indexing - 1D
print(a[1])
print(a[-3])

print(a[1:4])                          #slicing - [start:end:jump]
print(a[-3:-1])                   
print(a[::])    
print(a[::-3])                 
```

```python
a2=np.arange(9).reshape(3,3)                
print(a2)
```

```python
print(a2[1][1])                                   #indexing on 2D array to find 4,7,8 - row and columns start from index 0
print(a2[2,1])
print(a2[2,2])

print(a2[0,2])
```

```python
a2[2::,::]              #before comma there is row -start:end:skip and after it there is column
a2[0::2,::]     

#coulmn slicing
a2[0::2,1::]
a2[0::2,1::2]
```

```python
a2[2::,2::]          #slicing 8 from a2
a2[::,:1:]
```

```python
a3 =np.arange(27).reshape((3,3,3))                                #3D Array
print(a3)
print(a3.ndim)
```

```python
print(a3[0][0][0])                           #3D Indexing - [matrix no.(rotation) starts from 0 only] [row][column]
print(a3[1][0][0])
print(a3[1][0][1])

#find no. 5,17,21 from 3D Array a3

print(a3[0][1][2])
print(a3[1][2][2])
print(a3[2][1][0])
```

```python
a3[::,::,::]                             #3D Array Slicing
a3[1:2,::,::]
a3[0::2,::,::]
a3[1::2,1:2:,::]
a3[0::2 ,2::,::]

```
