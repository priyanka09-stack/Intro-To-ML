```python
from sklearn.impute import SimpleImputer
import pandas as pd
df =pd.read_csv('titanic.csv')
df['Age'].isnull().sum()
```

```python
imputer = SimpleImputer(strategy='mean')                 #mean
df['Age'] = imputer.fit_transform(df[['Age']])
df['Age'].isnull().sum()
```

```python
imputer1 = SimpleImputer(strategy='median')                        #median
df['Age'] = imputer1.fit_transform(df[['Age']])
```

```python
imputer2 = SimpleImputer(strategy='most_frequent')                        #mode
df['Age'] = imputer2.fit_transform(df[['Age']])
```

```python
imputer3 = SimpleImputer(strategy ='constant',fill_value=45)                          #Constant Value
df['Age'] = imputer3.fit_transform(df[['Age']])
```

```python
df['Embarked'].isnull().sum()
```

```python
imputer = SimpleImputer(strategy='most_frequent')                                  #Categorical - Mode
df[['Embarked']] = imputer.fit_transform(df[['Embarked']])

df['Embarked'].isnull().sum()
```

```python
imputer5 = SimpleImputer(strategy='constant',fill_value='unknown')                     #Categorical - Constant Value
df[['Embarked']] = imputer5.fit_transform(df[['Embarked']])
```

```python
df['Embarked'].isnull().sum()
```
