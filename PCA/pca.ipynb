```python
import pandas as pd
import numpy as np
```

```python
data = {
    'name': [
        'Aarav', 'Priya', 'Rahul', 'Ananya', 'Rohan',
        'Sneha', 'Karan', 'Ishita', 'Aditya', 'Meera'
    ],
    'maths': [85, 92, 65, 78, 55, 95, 72, 88, 60, 82],
    'physics': [78, 88, 70, 82, 60, 92, 68, 85, 55, 90],
    'chemistry': [82, 91, 68, 79, 58, 94, 75, 87, 62, 85],
    'english': [75, 85, 80, 88, 65, 90, 70, 92, 68, 95],
    'computer': [90, 95, 72, 85, 62, 98, 78, 91, 65, 88]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.drop('name',axis=1,inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler                                             # Feature Scaling
m = MinMaxScaler()
a = m.fit_transform(df)

np.round(a,2)
```

```python
a
```

```python
from sklearn.decomposition import PCA                                #PCA
pca = PCA(n_components=0.95)                                         #no./ percentage of feature to keep after PCA
dff = pca.fit_transform(a)
np.sum(pca.explained_variance_ratio_)
```
