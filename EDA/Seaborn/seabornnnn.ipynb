```python
#scatter plot
sns.scatterplot(data=df,x='total_bill', y='tip',hue='time',style='sex',size='size')            #numerical vs numerical , #hue - diff. on basis  of color,style - diff. on basis of structure ,size - diff. on basis of size
```

```python
sns.relplot(data=df,x='total_bill',y='tip',hue='time',style='sex',size='size')                 #Relational Plot
```

```python
marks=[30,40,50,60,70,80]                                           #LinePlot
days=['mon','tue','wed','thu','fri','sat']

sns.lineplot(x=days,y=marks)
```

```python
#Matrix Plot
'''
1. HeatMap
2.Cmap
'''
```

```python
#Heatmap
df1=pd.read_csv('data.csv')
df1
```

```python
heatmap_data = df1.set_index("City")                                     #HeatMap
sns.heatmap(
    heatmap_data,

    cmap='RdYlGn_r',
    cbar_kws={"label":'AQI'}
)
```

```python
df2 = pd.read_csv("titanic.csv")
df2
```

```python
#realtion between numeric values

correlation = df2.corr(numeric_only=True)                                          #Correlation
correlation
```

```python
sns.heatmap(                                                       #Correlation Heatmap
    correlation,
    annot=True,
    cbar_kws={"label":"Correlation Heatmap"}

    )
```

```python
'''
Categorical Plot
1.Strip Plot
2.Swarm Plot
3.Box Plot
4.Violin plot
5.Count plot
6.Bar plot

'''
```
