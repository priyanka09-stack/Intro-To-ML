```python
data=pd.read_csv("titanic.csv")
x=data["Age"]
```

```python
plt.hist(x)                   #histogram
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
```

```python
y=data["Fare"]
```

```python
plt.pie(y)
plt.title("Fare Distribution")
plt.show()
```

```python
marks=[3,5,8,5,0,3,57,9,32,1,1,5,7,9,11,15]           #histogram -- univariate analysis -- categorical data , continuous

plt.hist(marks,bins=[0,5,10,15,20],color='lightblue',label="Marks",edgecolor='black',orientation='vertical') #range -- 0-5|5-10|10-15  -- bins {assigns a range to tha data}

plt.legend()
plt.xlabel("MARKS")
plt.ylabel("FREQUENCY")
plt.title("Marks Distribution")
```

```python
marks=[20,30,10,20,20]
sub=["ML","DAA","DBMS","CC","CS"]

plt.pie(marks,labels=sub,autopct="%0.1f%%",colors=["navy",'pink','lightblue','green','cyan'],explode=[0.4,0,0,0.2,0])    #pie chart -- categorical and numerical data

```
